```python
#induced deliveries
table1 = pd.read_excel('Table1_induced.xlsx')
table1
```

```python
#figuring out which rows to drop
#finding rows that do not have an early record
i = 1
drop = []
while i < len(table1):
  if type(table1.loc[i,'Early record']) != int:
    drop.append(i)
  i += 1
print(drop)

#finding rows that do not have a later record
j= 1
while j < len(table1):
  if type(table1.loc[j,'Later record']) != int:
    drop.append(j)
  j += 1
print(drop)
```

```python
#updated table to only include data from people that had both early and late recordings
updated_induced = table1.drop(drop, axis=0)
updated_induced
```

```python
#removing data that is not relevant to our study
relevant_induced = updated_induced.drop(['Induced deliveries','Age','Weight (early)','Weight (later)','Height','Placental position','Newborn weight'], axis=1)
relevant_induced
```

```python
#Making a demonstration graph for the project proposal
folder = '/content/drive/MyDrive/copy of BMEN 600/BMEN 600/EHG Dataset/files/icehg-ds/1.0.1'
demo_induced_name = '/early_induced/icehg' + str(relevant_induced.iloc[1,0])
demo_induced_path = folder + demo_induced_name#file path for the first early induced recording
print(demo_induced_path)
```

```python
import wfdb
demo_record = wfdb.rdrecord(demo_induced_path)

print(demo_record.sig_name)#figuring out what order the signal information is given in
#S1, S1 filtered, S2, S2 filtered, S3, S3 filtered
print(demo_record.fs)#sampling frequency 20 Hz
print(demo_record.units)#getting the units of the signals
```

```python
# Extract the raw S1 signal
S1_signal = demo_record.p_signal[:, 0]
print(S1_signal)#testing that it worked
```

```python
# Create a time variable from the sampling frequency
# evenly spaced samples / frequency --> seconds. seconds / 60 --> minutes
import numpy as np
S1_time = np.arange(len(S1_signal)) / demo_record.fs / 60

print(S1_time) #testing that it worked
```

```python
# Raw S1 signal plot
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 4))

plt.plot(S1_time, S1_signal)

plt.xlabel('Time (minutes)')
plt.ylabel('Amplitude (mV)')
plt.title('Raw S1 EHG Signal - Early Induced (' + str(relevant_induced.iloc[1,1]) + ' weeks) ' + 'icehg' + str(relevant_induced.iloc[1,0]))
plt.grid()

plt.show()
```

```python
#Filtered S1 signal plot

filtered_S1_signal = demo_record.p_signal[:, 1] #0 last time, 1 is the filtered

plt.figure(figsize=(12, 4))

plt.plot(S1_time, filtered_S1_signal)

plt.xlabel('Time (minutes)')
plt.ylabel('Amplitude (mV)')
plt.title('Filtered S1 EHG Signal - Early Induced (' + str(relevant_induced.iloc[1,1]) + ' weeks) ' + 'icehg' + str(relevant_induced.iloc[1,0]))
plt.grid()

plt.show()
```

```python
#both on the same plot
plt.figure(figsize=(12, 4))
plt.plot(S1_time, S1_signal, label = 'Raw S1', color = 'black')
plt.plot(S1_time, filtered_S1_signal, label = 'Filtered S1', color = 'blue')

plt.xlabel('Time (minutes)')
plt.ylabel('Amplitude (mV)')
plt.title('S1 EHG Signal - Early Induced (' + str(relevant_induced.iloc[1,1]) + ' weeks) ' + 'icehg' + str(relevant_induced.iloc[1,0]))
plt.legend()
plt.grid()
plt.show()
```

```python
#Removing the start and end to remove the amplitude spike at the start
#Since electrode placement and removal could mess with the recording, both the start and end were taken off
plt.figure(figsize=(12, 4))
plt.plot(S1_time, S1_signal, label = 'Raw S1', color = 'black')
plt.plot(S1_time, filtered_S1_signal, label = 'Filtered S1', color = 'blue')

plt.xlim(2.5,S1_time[-1] - 2.5) #removing the first and last 2.5 minutes (150 seconds)
#The same amount of time will be taken off of all the recordings
plt.xlabel('Time (minutes)')
plt.ylabel('Amplitude (mV)')
plt.title('S1 EHG Signal - Early Induced (' + str(relevant_induced.iloc[1,1]) + ' weeks) ' + 'icehg' + str(relevant_induced.iloc[1,0]))
plt.legend()
plt.grid()
plt.show()
```
